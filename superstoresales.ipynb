Questions

```python
category_sales = df.groupby('Category').agg(Total_Sales = ('Sales' , 'sum') , 
                           Total_Profit = ('Profit' , 'sum'),
                           Total_Quantity = ('Quantity' , 'sum')).sort_values('Total_Profit' , ascending= False)
```

```python
category_sales['Profit_margin'] = category_sales['Total_Profit']/ category_sales['Total_Sales'] * 100
```

```python
category_sales
```

```python
category_sales.plot(kind='bar'  , y =['Total_Sales' , 'Total_Profit'] , color = ['navy' , 'green'])
for i , value in enumerate(category_sales['Total_Sales']):
    plt.text(i , value , f'{int(value/1000)}K' , ha = 'right' , va = 'bottom')
    
for i , value in enumerate(category_sales['Total_Profit']):
    plt.text(i+0.2, value , f'{int(value/1000)}K', ha = 'right' , va = 'bottom')
    
plt.xlabel("")
plt.title('Profit and Sales by Each Category')
```

```python
category_sales.plot(kind ='pie' , y = 'Total_Profit' , autopct='%1.1f%%')
plt.ylabel('')
plt.legend().remove()
plt.title('Profit Share by Each Product Category')
```

In the Bar chart, we can see that Office Supplies is the best selling product category. 

But its profit is lower than the profit from Technology product. 

This means that Technological product are the most profit generating product for our store.

The profit margin of Technology category is 19 % which mean we can get $ 19 whenever 

we sell $ 100 Technology product.

The worst Category by profit is Furniture. We get total about $ 10000 from this category.

In profit share, `51.6 % ` of whole profit comes from Technologly Category , `42.7%` comes from Offive Supplies Category and 

` 5.7 % ` comes from Furniture Category.

```python
df.info()
```

```python
df['Month_name'] = df['Order Date'].dt.strftime('%b')
```

```python
monthly_sales= df.groupby('Month_name').agg(Total_Sales = ('Sales' , 'sum'),
                             Total_Profit = ('Profit' , 'sum'))
```

```python
monthly_sales.reset_index(inplace= True)
```

```python
monthly_sales['Month_no'] = pd.to_datetime(monthly_sales['Month_name'] , format=('%b')).dt.month
```

```python
monthly_sales.sort_values('Month_no' , ascending= True , inplace= True)
```

```python
monthly_sales
```

```python
sns.lineplot(monthly_sales , x = 'Month_name' , y = 'Total_Sales' , marker = 'o' , 
             markeredgecolor = 'black' , markersize = 4)

plt.ylabel('')

plt.title('Monthly Sales Analysis')
```

```python
trend_df = df.copy()
```

```python
trend_df['Period'] = trend_df['Order Date'].dt.to_period('M')
```

```python
period_sales = trend_df.groupby('Period').agg(Total_Sale = ('Sales' ,'sum'),
                                              Total_Profit = ('Profit' , 'sum')).reset_index()
```

```python
period_sales['Month_name'] = period_sales['Period'].dt.strftime('%b')
```

```python
period_sales.info()
```

```python
fig , ax = plt.subplots(figsize = (12 , 5))

period_sales.plot(kind= 'line' , x = 'Period' , y = 'Total_Sale' , ax = ax , marker = 'o',color = 'navy')

ax.set_xlabel('')
ax.legend().remove()
ax.set_title('Monthly Sales Growth')
```

```python
fig , ax = plt.subplots(figsize = (12 , 5))

period_sales.plot(kind= 'line' , x = 'Period' , y = 'Total_Sale' , ax = ax , marker = 'o',color = 'gray')

for i, (x_val, value) in enumerate(zip(period_sales['Period'], period_sales['Total_Sale'])):
    ax.text(x_val, value, f'{int(value/1000)}K' , ha = 'left' , va = 'bottom' , color = 'darkgreen')
ax.set_xlabel('')
ax.legend().remove()
ax.set_title('Monthly Sales Growth')
```

Above Line Chart show that our sale are strongly seasonal effect.

Sales are slightly uprising in Jan and Feb. 

In April and October, our sales are always downed.

In Q1 March is the highest and our sales is downed again in April.
